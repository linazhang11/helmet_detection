```python
import os

def create_project_structure():
    """创建安全帽检测项目的文件夹结构"""
    
    # 项目根目录 - 桌面上的helmet_detection_project文件夹
    desktop_path = os.path.join(os.path.expanduser("~"), "Desktop")
    project_root = os.path.join(desktop_path, "helmet_detection_project")
    
    # 定义文件夹结构
    folders = [
        # 数据集目录
        "dataset/images/train",
        "dataset/images/val", 
        "dataset/labels/train",
        "dataset/labels/val",
        
        # 原始数据目录
        "data/raw",
        
        # 结果目录
        "results/inference",
        
        # 模型目录
        "models",
        
        # Jupyter notebook目录
        "notebooks",
        
        # 工具函数目录
        "utils",
        
        # Jupyter缓存（通常会自动生成，但也可以手动创建）
        ".ipynb_checkpoints",
        
        # 训练输出目录（通常会自动生成）
        "runs"
    ]
    
    # 定义需要创建的文件
    files = [
        "dataset/dataset.yaml",
        "dataset/classes.txt",
        "data/raw/test_image.jpg"
    ]
    
    print(f"在桌面创建项目文件夹: {project_root}")
    
    # 创建所有文件夹
    for folder in folders:
        folder_path = os.path.join(project_root, folder)
        os.makedirs(folder_path, exist_ok=True)
        print(f"✓ 创建文件夹: {folder_path}")
    
    # 创建所有文件（空文件）
    for file in files:
        file_path = os.path.join(project_root, file)
        # 确保父目录存在
        os.makedirs(os.path.dirname(file_path), exist_ok=True)
        # 创建空文件
        with open(file_path, 'w') as f:
            pass
        print(f"✓ 创建文件: {file_path}")
    
    print(f"\n项目结构创建完成！")
    print(f"项目位置: {project_root}")
    
    # 显示完整的文件夹结构
    print(f"\n生成的文件夹结构:")
    display_folder_structure(project_root)

def display_folder_structure(startpath):
    """显示文件夹结构"""
    for root, dirs, files in os.walk(startpath):
        level = root.replace(startpath, '').count(os.sep)
        indent = ' ' * 2 * level
        print(f'{indent}📁 {os.path.basename(root)}/')
        subindent = ' ' * 2 * (level + 1)
        for file in files:
            print(f'{subindent}📄 {file}')

# 执行创建函数
create_project_structure()
```
